```python
import phonenumbers
import opencage
import folium
from phonenumbers import geocoder
from phonenumbers import carrier
from opencage.geocoder import OpenCageGeocode
```

```python


number = "6379613654"  # Your phone number
region = "IN"  # Assuming this is an Indian number

pepnumber = phonenumbers.parse(number,region)
location = geocoder.description_for_number(pepnumber, "en")
```

```python
service_pro = phonenumbers.parse(number,region)
print(location)
print(carrier.name_for_number(service_pro,"en"))
```

```python
key = "aaa320a4296d42e68fc77b6aa7f94ea6"
query = str(location)
geocoder = OpenCageGeocode(key)
results = geocoder.geocode(query)
print(results)
```

```python
lat = results[0]['geometry']['lat'] 
lng = results[0]['geometry']['lng'] 
print(lat,lng)
```

```python
myMap = folium.Map(location = [lat,lng],zoom_start = 9)
folium.Marker([lat,lng],popup = location).add_to(myMap)
myMap.save("mylocation.html")
```
